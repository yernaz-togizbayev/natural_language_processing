Extend the chunk grammar from Exercise 1 to a multi-stage chunk grammar for a cascaded chunker that produces the following output for the three test sentences:

    (S
      (NP many/JJ dogs/NNS)
      (VP barked/VBD (PP at/IN (NP the/DT cat/NN))))
    (S
      (NP two/CD dogs/NNS)
      (VP barked/VBD (PP at/IN (NP the/DT cat/NN))))
    (S
      (NP both/DT new/JJ dogs/NNS)
      (VP barked/VBD (PP at/IN (NP the/DT cat/NN))))

```python
new_sentence1 = [("many", "JJ"), ("dogs", "NNS"), ("barked", "VBD"), ("at", "IN"), ("the", "DT"), ("cat", "NN")]
new_sentence2 = [("two", "CD"), ("dogs", "NNS"), ("barked", "VBD"), ("at", "IN"), ("the", "DT"), ("cat", "NN")]
new_sentence3 = [("both", "DT"), ("new", "JJ"), ("dogs", "NNS"), ("barked", "VBD"), ("at", "IN"), ("the", "DT"), ("cat", "NN")]

new_sentences = [new_sentence1, new_sentence2, new_sentence3]

grammar = r"""
            NP: {<DT|CD>?<JJ>*<NN|NNS>}
            PP: {<IN><NP>}
            VP: {<VB.*><NP|PP>}
          """

new_cp = nltk.RegexpParser(grammar)

new_results = [new_cp.parse(sent) for sent in new_sentences]

for i, sent in enumerate(new_results, start=1):
    print(f"\nChunked Sentence {i}:")
    print(sent.pprint())
    print('---' * 10)
```

### Exercise 7

Select the VP chunks from the "train" portion of the CoNLL 2000 Chunking Data Corpus. Create a list of all the tag sequences that occur with each instance of this chunk type. Use a `FreqDist` to display the 100 most common tag sequences.  

```python
train_sents = conll2000.chunked_sents('train.txt', chunk_types=['VP'])

def extract_tag_sequences(chunked_sents):
    tag_sequences = []
    for sent in chunked_sents:
        for subtree in sent:
            if isinstance(subtree, nltk.Tree) and subtree.label() == 'VP':
                tags = [tag for _, tag in subtree.leaves()]
                tag_sequences.append(" ".join(tags))
    return tag_sequences

vp_tag_sequences = extract_tag_sequences(train_sents)

# Frequency distribution
fdist = FreqDist(vp_tag_sequences)

# 100 most common
most_common_vp_tag_sequences = fdist.most_common(100)

for seq, freq in most_common_vp_tag_sequences:
    print(f"{seq}: {freq}")
```
